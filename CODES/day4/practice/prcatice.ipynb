```python
import pandas as pd
```

```python
dict_a = pd.DataFrame({'name' : ['anup', ' ram', 'hari'],
                      'roll no' : [7, 8 ,9],
                      'marks': [600,700,800]})
dict_b = pd.DataFrame({'name' : ['shyam', 'jhyam', 'lyam'],
                      'roll no' : [77, 88 ,99]})
pd.concat([dict_a,dict_b], axis = 0)
```

```python
students = pd.DataFrame({
    'std_id':[2,3,4,5],
    'name': ['anup', 'kushal', 'aananta', 'dhiraj']
})

results = pd.DataFrame({
    'std_id':[1,2,3,4,5],
    'marks': [100,200,300,400,500],
    'gpa': [3.14,3.50,3.80,4.00,2.00]
})

pd.merge(students, results, on= 'std_id', how='right', indicator = True)
```

```python
left = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
right = pd.DataFrame({'user_id': [1, 2, 4], 'score': [90, 85, 70]})
pd.merge(left, right, left_on='id', right_on='user_id', how='left')
```

- here the primary keys are different in both the tables and to tell the system
- which keys from left and which keys from the right to work on we use the left_on and right_on
